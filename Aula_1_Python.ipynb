Desafio crie um código que percorra a lista e some 1 ponto, exceto quando o valor passar de 10, nota máxima

```python
n = 0
while n < 5:
  if lista_de_medias[n] + 1 <= 10:
    lista_de_medias[n] = lista_de_medias[n] + 1
  else:
    lista_de_medias[n] = 10
  n += 1
print(lista_de_medias)
```
