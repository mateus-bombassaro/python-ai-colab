Desafio:
1. Pedir para IA gerar em python uma lista de 20 emails
2. Criar função que vai receber como parâmetro uma lista de emails, percorrer, para cada email vai chamar a IA do Gemini resumir o email em apenas uma linha, definindo o que essa pessoa queria no email.

```python
def email_resume(emails_list):
  lista_de_resumos = []
  for numero, email in enumerate(emails_list):
    resposta = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=f"""Vou te mandar o corpo de um email, quero que você o
        resuma em apenas 1 linha, segue o email: {email}"""
      )
    print(f"Email {numero + 1}: {resposta.text}")
    lista_de_resumos.append(f"Email {numero + 1}: {resposta.text}")
    print("-" * 50)

  return lista_de_resumos
```

```python
emails_corpo = [
    # 1. Email Comercial
    """Prezado(a) [Nome],

Espero que esteja bem. Entrei em contato para apresentar nossos serviços que podem contribuir significativamente para o crescimento do seu negócio. Possuímos uma equipe experiente pronta para atender suas necessidades.

Fico à disposição para agendar uma reunião e discutir como podemos colaborar.

Atenciosamente,
[Seu Nome]""",

    # 2. Email de Oportunidade de Emprego
    """Olá [Nome],

Vimos seu currículo e ficamos impressionados com sua experiência. Gostaríamos de convidá-lo(a) para participar do processo de seleção para a posição de [Cargo] em nossa empresa.

Por favor, confirme sua disponibilidade para uma entrevista preliminar.

Aguardamos seu retorno.

Atenciosamente,
Departamento de Recursos Humanos""",

    # 3. Email de Confirmação de Pedido
    """Caro(a) [Nome],

Agradecemos seu pedido! Confirmamos que recebemos seu pedido nº [Número] realizado em [Data].

Seu pedido será processado e você receberá atualizações sobre o status do envio em breve.

Obrigado pela preferência!

Atenciosamente,
Equipe de Vendas""",
]

lista_resumos = email_resume(emails_corpo)
print(lista_resumos)
```

```python
from groq import Groq

client = Groq()
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": ""
      }
    ],
    temperature=1,
    max_completion_tokens=8192,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")
```

# Manipular dados com Panda

```python
import numpy as np

# Gerar 50 linhas de dados
n_rows = 50

# Nomes de produtos aleatórios
product_names = [f'Produto {i+1}' for i in range(n_rows)]

# Categorias de produtos
categories = ['Eletrônicos', 'Alimentos', 'Vestuário', 'Livros', 'Casa']
product_categories = np.random.choice(categories, n_rows)

# Preços aleatórios entre 10 e 1000
prices = np.round(np.random.uniform(10, 1000, n_rows), 2)

# Quantidade de produtos vendidos aleatória entre 1 e 100
quantities_sold = np.random.randint(1, 101, n_rows)

# Criar o DataFrame
df_produtos = pd.DataFrame({
    'Nome do produto': product_names,
    'Categoria do produto': product_categories,
    'Preço': prices,
    'Quantos produtos foram vendidos': quantities_sold
})

display(df_produtos.head())
```
